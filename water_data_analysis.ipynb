```python
Q1 = data["Annual_Rainfall_mm"].quantile(0.25)
Q3 = data["Annual_Rainfall_mm"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

```python
outliers = data[
    (data["Annual_Rainfall_mm"] < lower_bound) |
    (data["Annual_Rainfall_mm"] > upper_bound)
]

outliers
```

```python
train_data = data[data['Year'] <= 2023]
test_data = data[data['Year'] >= 2024]

X_train = train_data[['Mean_Temperature_C']]
y_train = train_data['Annual_Rainfall_mm']

X_test = test_data[['Mean_Temperature_C']]
y_test = test_data['Annual_Rainfall_mm']

print("Train:", train_data['Year'].tolist())
print("Test:", test_data['Year'].tolist())
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
y_test_pred = model.predict(X_test)

from sklearn.metrics import mean_absolute_error, r2_score

mae = mean_absolute_error(y_test, y_test_pred)
r2 = r2_score(y_test, y_test_pred)

print("Test MAE:", mae)
print("Test R²:", r2)
```

```python
plt.plot(test_data['Year'], y_test, marker='o', label='Actual')
plt.plot(test_data['Year'], y_test_pred, marker='o', label='Predicted')

plt.xlabel('Year')
plt.ylabel('Annual Rainfall (mm)')
plt.title('Actual vs Predicted Rainfall')
plt.legend()
plt.show()
```
